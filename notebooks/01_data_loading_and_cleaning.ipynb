```python
from ucimlrepo import fetch_ucirepo

dataset = fetch_ucirepo(id=697)

X = dataset.data.features
y = dataset.data.targets
```

```python
df["Course"].value_counts()
```

```python
course_outcomes = pd.crosstab(
    df["Course"],
    df["Target"],
    normalize="index"
) * 100

course_outcomes
```

```python
course_dropout = course_outcomes["Dropout"].sort_values(
    ascending=False
)

course_dropout
```

```python
plt.figure(figsize=(10, 7))

sns.barplot(
    x=course_dropout.values,
    y=course_dropout.index.astype(str),
    orient="h"
)

plt.title("Dropout Rate by Course")
plt.xlabel("Dropout Rate (%)")
plt.ylabel("Course")

plt.show()
```

```python
course_summary = pd.DataFrame({
    "Students": df["Course"].value_counts(),
    "Dropout Rate (%)": course_dropout
})
```

```python
course_summary.sort_values(
    "Dropout Rate (%)",
    ascending=False
)
```

```python
scholarship_outcomes = pd.crosstab(
    df["Scholarship holder"],
    df["Target"],
    normalize="index"
) * 100

scholarship_outcomes
```

```python
tuition_outcomes = pd.crosstab(
    df["Tuition fees up to date"],
    df["Target"],
    normalize="index"
) * 100

tuition_outcomes
```

```python
scholarship_outcomes.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Student Outcome by Scholarship Status")
plt.xlabel("Scholarship Holder")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
tuition_outcomes.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Student Outcome by Tuition Fee Status")
plt.xlabel("Tuition Fees Up to Date")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
scholarship_dropout = (
    df.groupby("Scholarship holder")["Target"]
    .apply(lambda x: (x == "Dropout").mean() * 100)
)

scholarship_dropout
```

```python
tuition_dropout = (
    df.groupby("Tuition fees up to date")["Target"]
    .apply(lambda x: (x == "Dropout").mean() * 100)
)

tuition_dropout
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    x=scholarship_dropout.index.astype(str),
    y=scholarship_dropout.values
)

plt.title("Dropout Rate by Scholarship Status")
plt.xlabel("Scholarship Holder (0 = No, 1 = Yes)")
plt.ylabel("Dropout Rate (%)")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    x=tuition_dropout.index.astype(str),
    y=tuition_dropout.values
)

plt.title("Dropout Rate by Tuition Fee Status")
plt.xlabel("Tuition Fees Up to Date (0 = No, 1 = Yes)")
plt.ylabel("Dropout Rate (%)")

plt.show()
```
